# Conway Kochen 55 — Independence number

$COMPLETED$ 55- $CONWAY-KOCHEN SET$: $\theta$=25 $\alpha$=25 d=3

```python
import numpy as np
import itertools
import networkx as nx
vectors = [
    np.array([-1,  2,  1]),
    np.array([-1,  2,  0]),
    np.array([ 0,  2,  1]),
    np.array([-1,  2, -1]),
    np.array([ 0,  2,  0]),
    np.array([ 1,  2,  1]),
    np.array([ 0,  2, -1]),
    np.array([ 1,  2,  0]),
    np.array([ 1,  2, -1]),
    np.array([ 0,  2, -2]),
    np.array([ 2,  2,  0]),
    np.array([ 2,  2, -2]),
    np.array([-1,  1, -2]),
    np.array([ 0,  1, -2]),
    np.array([-1,  0, -2]),
    np.array([ 0, 0, -2]),
    np.array([-1, -1, -2]),
    np.array([ 0, -1, -2]),
    np.array([ 0, -2, -2]),
    np.array([ 2,  1, -1]),
    np.array([ 2,  1,  0]),
    np.array([ 2,  0, -1]),
    np.array([ 2,  0,  0]),
    np.array([ 2, -1, -1]),
    np.array([ 2, -1,  0]),
    np.array([ 2, -2,  0]),
    np.array([ 2,  0, -2]),
    np.array([ 2, -2, -2]),
    np.array([ 2,  2,  2]),
    np.array([ 2, -2,  2]),
    np.array([ 2,  0,  2]),
    np.array([ 4, -4, -8]),
    np.array([-5, -1,  2]),
    np.array([-4, -4,  8]),
    np.array([-1, -2, -5]),
    np.array([ 5,  2, -1]),
    np.array([-1, -5, -2]),
    np.array([ 1, -5,  2]),
    np.array([-1, -2,  5]),
    np.array([-8, -4, -4]),
    np.array([ 5, -1, -2]),
    np.array([-5,  2, -1]),
    np.array([ 2,  5, -1]),
    np.array([-2,  1, -5]),
    np.array([ 1, -2, -5]),
    np.array([-2, -1, -5]),
    np.array([-5, -2, -1]),
    np.array([-5,  2,  1]),
    np.array([ 8, -4,  4]),
    np.array([-1,  2, -5]),
    np.array([-2,  5,  1]),
    np.array([20,  4,  8]),
    np.array([-20,  4, -8]),
    np.array([ 8,  4, -20]),
    np.array([ 8, -4, -20]),

]
n = len(vectors)
d = len(vectors[0])


def is_orthogonal(v1, v2, tol=1e-9):
    return abs(np.dot(v1, v2)) < tol

G = nx.Graph()
G.add_nodes_from(range(n))
for i, j in itertools.combinations(range(n), 2):
    if is_orthogonal(vectors[i], vectors[j]):
        G.add_edge(i, j)

maximal_cliques = list(nx.find_cliques(G))
hyperedges = [c for c in maximal_cliques if len(c) == d]

print(f"Orthogonality graph: {n} vertices, {G.number_of_edges()} edges")
print(f"Hyperedges (full {d}-dim orthogonal bases): {len(hyperedges)}")


complement = nx.complement(G)
max_clique = max(nx.find_cliques(complement), key=len)
alpha = len(max_clique)

print(f"\nIndependence number alpha(H) = {alpha}")
print("A maximum independent set (1-based indices):",
      sorted(i + 1 for i in max_clique))
print(f"alpha * d = {alpha} * {d} = {alpha * d}")

print("\nHyperedges (orthogonal bases), 1-based indices:")
for e in sorted(hyperedges, key=lambda c: sorted(c)):
    print(" ", sorted(i + 1 for i in e))
```
