```python
import numpy as np
import itertools
import cvxpy as cp
vectors = [
    np.array([0,2,0,0,0,0,0,0]),
    np.array([0,0,2,0,0,0,0,0]),
    np.array([0,0,0,2,0,0,0,0]),
    np.array([0,0,0,0,2,0,0,0]),
    np.array([0,0,0,0,0,2,0,0]),
    np.array([0,0,0,0,0,0,2,0]),
    np.array([0,0,0,0,0,0,0,2]),
    np.array([1,1,1,1,0,0,0,0]),
    np.array([1,1,-1,-1,0,0,0,0]),
    np.array([1,-1,1,-1,0,0,0,0]),
    np.array([1,-1,-1,1,0,0,0,0]),
    np.array([0,0,0,0,1,1,-1,-1]),
    np.array([0,0,0,0,1,-1,1,-1]),
    np.array([0,0,0,0,1,-1,-1,1]),
    np.array([1,1,0,0,1,1,0,0]),
    np.array([1,1,0,0,-1,-1,0,0]),
    np.array([1,-1,0,0,1,-1,0,0]),
    np.array([1,-1,0,0,-1,1,0,0]),
    np.array([0,0,1,1,0,0,1,1]),
    np.array([0,0,1,1,0,0,-1,-1]),
    np.array([0,0,1,-1,0,0,-1,1]),
    np.array([1,0,1,0,1,0,1,0]),
    np.array([1,0,1,0,-1,0,-1,0]),
    np.array([1,0,-1,0,1,0,-1,0]),
    np.array([0,1,0,1,0,1,0,1]),
    np.array([0,1,0,1,0,-1,0,-1]),
    np.array([0,1,0,-1,0,1,0,-1]),
    np.array([0,1,0,-1,0,-1,0,1]),
    np.array([1,0,0,1,0,1,-1,0]),
    np.array([1,0,0,-1,0,1,1,0]),
    np.array([1,0,0,1,0,-1,1,0]),
    np.array([1,0,0,-1,0,-1,-1,0]),
    np.array([0,1,1,0,-1,0,0,1]),
    np.array([0,1,-1,0,1,0,0,1]),
    np.array([0,-1,1,0,1,0,0,1]),
    np.array([0,-1,-1,0,-1,0,0,-1]),

]

n = len(vectors)
print("n =", n)

def hermitian_inner(u, v):
    return np.vdot(u, v)

tol = 1e-9
adj = np.zeros((n, n), dtype=int)
for i, j in itertools.combinations(range(n), 2):
    if abs(hermitian_inner(vectors[i], vectors[j])) < tol:
        adj[i, j] = adj[j, i] = 1

X = cp.Variable((n, n), symmetric=True)
theta = cp.Variable()

constraints = [X >> 0]

for i in range(n):
    constraints.append(X[i, i] == theta - 1)

for i in range(n):
    for j in range(i+1, n):
        if adj[i, j] == 0:
            constraints.append(X[i, j] == -1)
prob = cp.Problem(cp.Minimize(theta), constraints)

print("Solving SDP ... (may take some time)")
res = prob.solve(solver=cp.SCS, verbose=True, eps=1e-6, max_iters=200000)
print("Optimal theta ≈", res)
```
