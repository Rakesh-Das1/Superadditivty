# YO 25 — Independence number

YO-25 $\theta(G)$=13 $\alpha$=13 d=3

```python
import numpy as np
import itertools
import networkx as nx
vectors = [
    np.array([1, 0, 0]),#1
    np.array([0, 1, 0]),#2
    np.array([0, 0, 1]),#3
    np.array([1, 1, 1]) / np.sqrt(3),#4
    np.array([1, 1, -1]) / np.sqrt(3),#5
    np.array([1, -1, 1]) / np.sqrt(3),#6
    np.array([-1, 1, 1]) / np.sqrt(3),#7
    np.array([0, 1, 1]) / np.sqrt(2),#8
    np.array([0, 1, -1]) / np.sqrt(2),#9
    np.array([1, 0, 1]) / np.sqrt(2),#10
    np.array([1, 0, -1]) / np.sqrt(2),#11
    np.array([1, 1, 0]) / np.sqrt(2),#12
    np.array([1, -1, 0]) / np.sqrt(2),#13
    np.array([2,-1,-1]),#14
    np.array([1,1,-2]),#15
    np.array([-1,2,-1]),#16
    np.array([-2,-1,-1]),#17
    np.array([1,2,-1]),#18
    np.array([-1,1,-2]),#19
    np.array([1,2,1]),#20
    np.array([-1,1,2]),#21
    np.array([-2,-1,1]),#22
    np.array([2,-1,1]),#23
    np.array([-1,-1,-2]),#24
    np.array([1,-2,-1])#25
]

n = len(vectors)
d = len(vectors[0])


def is_orthogonal(v1, v2, tol=1e-9):
    return abs(np.dot(v1, v2)) < tol

G = nx.Graph()
G.add_nodes_from(range(n))
for i, j in itertools.combinations(range(n), 2):
    if is_orthogonal(vectors[i], vectors[j]):
        G.add_edge(i, j)

maximal_cliques = list(nx.find_cliques(G))
hyperedges = [c for c in maximal_cliques if len(c) == d]

print(f"Orthogonality graph: {n} vertices, {G.number_of_edges()} edges")
print(f"Hyperedges (full {d}-dim orthogonal bases): {len(hyperedges)}")


complement = nx.complement(G)
max_clique = max(nx.find_cliques(complement), key=len)
alpha = len(max_clique)

print(f"\nIndependence number alpha(H) = {alpha}")
print("A maximum independent set (1-based indices):",
      sorted(i + 1 for i in max_clique))
print(f"alpha * d = {alpha} * {d} = {alpha * d}")

print("\nHyperedges (orthogonal bases), 1-based indices:")
for e in sorted(hyperedges, key=lambda c: sorted(c)):
    print(" ", sorted(i + 1 for i in e))
```
