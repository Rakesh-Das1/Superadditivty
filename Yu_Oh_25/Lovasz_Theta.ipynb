```python
import numpy as np
import itertools
import cvxpy as cp
vectors = [
    np.array([1, 0, 0]),#1
    np.array([0, 1, 0]),#2
    np.array([0, 0, 1]),#3
    np.array([1, 1, 1]) / np.sqrt(3),#4
    np.array([1, 1, -1]) / np.sqrt(3),#5
    np.array([1, -1, 1]) / np.sqrt(3),#6
    np.array([-1, 1, 1]) / np.sqrt(3),#7
    np.array([0, 1, 1]) / np.sqrt(2),#8
    np.array([0, 1, -1]) / np.sqrt(2),#9
    np.array([1, 0, 1]) / np.sqrt(2),#10
    np.array([1, 0, -1]) / np.sqrt(2),#11
    np.array([1, 1, 0]) / np.sqrt(2),#12
    np.array([1, -1, 0]) / np.sqrt(2),#13
    np.array([2,-1,-1]),#14
    np.array([1,1,-2]),#15
    np.array([-1,2,-1]),#16
    np.array([-2,-1,-1]),#17
    np.array([1,2,-1]),#18
    np.array([-1,1,-2]),#19
    np.array([1,2,1]),#20
    np.array([-1,1,2]),#21
    np.array([-2,-1,1]),#22
    np.array([2,-1,1]),#23
    np.array([-1,-1,-2]),#24
    np.array([1,-2,-1])#25
]
n = len(vectors)
print("n =", n)

def hermitian_inner(u, v):
    return np.vdot(u, v)

tol = 1e-9
adj = np.zeros((n, n), dtype=int)
for i, j in itertools.combinations(range(n), 2):
    if abs(hermitian_inner(vectors[i], vectors[j])) < tol:
        adj[i, j] = adj[j, i] = 1

X = cp.Variable((n, n), symmetric=True)
theta = cp.Variable()

constraints = [X >> 0]

for i in range(n):
    constraints.append(X[i, i] == theta - 1)

for i in range(n):
    for j in range(i+1, n):
        if adj[i, j] == 0:
            constraints.append(X[i, j] == -1)

prob = cp.Problem(cp.Minimize(theta), constraints)

print("Solving SDP ... (may take some time)")
res = prob.solve(solver=cp.SCS, verbose=True, eps=1e-6, max_iters=200000)
print("Optimal theta ≈", res)
```
