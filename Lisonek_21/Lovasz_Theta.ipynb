```python
import numpy as np
import itertools
import cvxpy as cp
w = np.exp(2j * np.pi / 3)
vectors = [
    np.array([1,0,0,0,0,0]),#1
    np.array([0,1,0,0,0,0]),#2
    np.array([0,0,1,0,0,0]),#3
    np.array([0,0,0,1,0,0]),#4
    np.array([0,0,0,0,1,0]),#5
    np.array([0,0,0,0,0,1]),#6
    np.array([0,0,1,1,1,1]),#7
    np.array([0,1,0,1,w,w**2]),#8
    np.array([0,1,1,0,w**2,w]),#9
    np.array([0,1,w,w**2,0,1]),#10
    np.array([0,1,w**2,w,1,0]),#11
    np.array([1,0,0,1,w**2,w]),#12
    np.array([1,0,1,0,w,w**2]),#13
    np.array([1,0,w**2,w,0,1]),#14
    np.array([1,0,w,w**2,1,0]),#15
    np.array([1,1,0,0,1,1]),#16
    np.array([w,w**2,0,1,0,1]),#17
    np.array([w**2,w,0,1,1,0]),#18
    np.array([w**2,w,1,0,0,1]),#19
    np.array([w,w**2,1,0,1,0]),#20
    np.array([1,1,1,1,0,0]),#21
]
n = len(vectors)
print("n =", n)

def hermitian_inner(u, v):
    return np.vdot(u, v)

tol = 1e-9
adj = np.zeros((n, n), dtype=int)
for i, j in itertools.combinations(range(n), 2):
    if abs(hermitian_inner(vectors[i], vectors[j])) < tol:
        adj[i, j] = adj[j, i] = 1

X = cp.Variable((n, n), symmetric=True)
theta = cp.Variable()

constraints = [X >> 0]

for i in range(n):
    constraints.append(X[i, i] == theta - 1)

# X_ij = -1 when i != j and i is NOT adjacent (i.e. non-edge)
for i in range(n):
    for j in range(i+1, n):
        if adj[i, j] == 0:
            constraints.append(X[i, j] == -1)

# objective
prob = cp.Problem(cp.Minimize(theta), constraints)

print("Solving SDP ... (may take some time)")
res = prob.solve(solver=cp.SCS, verbose=True, eps=1e-6, max_iters=200000)
print("Optimal theta ≈", res)
```
